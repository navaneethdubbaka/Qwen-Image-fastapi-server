```python
from pathlib import Path
import os

base = Path("/kaggle/working/ComfyUI")

for folder in ["diffusion_models", "text_encoders", "vae"]:
    (base / "models" / folder).mkdir(parents=True, exist_ok=True)

links = {
    "diffusion_models/qwen_image_2.1_turbo_Q4_K_S.gguf":
        "/kaggle/working/qwen_image_2.1_turbo_Q4_K_S.gguf",

    "text_encoders/qwen3vl_8b_w4a8.safetensors":
        "/kaggle/working/qwen3vl_8b_w4a8.safetensors",

    "vae/qwen_image_2.1_vae_bf16.safetensors":
        "/kaggle/working/qwen_image_2.1_vae_bf16.safetensors",
}

for relative, target in links.items():
    link = base / "models" / relative
    if link.exists() or link.is_symlink():
        link.unlink()
    os.symlink(target, link)

print("Models linked.")
```

```python
app_code = r'''
import json
import time
import random
import urllib.request
import urllib.parse
import urllib.error

from fastapi import FastAPI, HTTPException
from fastapi.responses import Response
from pydantic import BaseModel

COMFY = "http://127.0.0.1:8188"

app = FastAPI(title="Viggle Qwen Image API")


class GenerateRequest(BaseModel):
    prompt: str


def generate_image(prompt: str):

    workflow = {
        "1": {
            "class_type": "UnetLoaderGGUF",
            "inputs": {
                "unet_name": "qwen_image_2.1_turbo_Q4_K_S.gguf"
            }
        },

        "2": {
            "class_type": "CLIPLoader",
            "inputs": {
                "clip_name": "qwen3vl_8b_w4a8.safetensors",
                "type": "qwen_image",
                "device": "default"
            }
        },

        "3": {
            "class_type": "VAELoader",
            "inputs": {
                "vae_name": "qwen_image_2.1_vae_bf16.safetensors"
            }
        },

        "4": {
            "class_type": "TextEncodeQwenImage21",
            "inputs": {
                "clip": ["2", 0],
                "vae": ["3", 0],
                "prompt": prompt,
                "negative_prompt": "",
                "resolution": 1024
            }
        },

        "5": {
            "class_type": "KSampler",
            "inputs": {
                "model": ["1", 0],
                "positive": ["4", 0],
                "negative": ["4", 1],
                "latent_image": ["4", 2],
                "seed": random.randint(0, 2**32 - 1),
                "steps": 4,
                "cfg": 1.0,
                "sampler_name": "euler",
                "scheduler": "simple",
                "denoise": 1.0
            }
        },

        "6": {
            "class_type": "VAEDecode",
            "inputs": {
                "samples": ["5", 0],
                "vae": ["3", 0]
            }
        },

        "7": {
            "class_type": "SaveImage",
            "inputs": {
                "images": ["6", 0],
                "filename_prefix": "viggle_api"
            }
        }
    }

    # Send workflow to ComfyUI
    data = json.dumps({
        "prompt": workflow
    }).encode()

    request = urllib.request.Request(
        f"{COMFY}/prompt",
        data=data,
        headers={"Content-Type": "application/json"}
    )

    try:
        response = urllib.request.urlopen(request)
        result = json.loads(response.read())

    except urllib.error.HTTPError as e:
        error_body = e.read().decode()
        raise HTTPException(
            status_code=500,
            detail=f"ComfyUI error: {error_body}"
        )

    prompt_id = result["prompt_id"]

    # Wait for generation
    while True:

        time.sleep(2)

        try:
            with urllib.request.urlopen(
                f"{COMFY}/history/{prompt_id}"
            ) as response:

                history = json.loads(response.read())

        except Exception as e:
            raise HTTPException(
                status_code=500,
                detail=f"Failed reading ComfyUI history: {str(e)}"
            )

        if prompt_id in history:
            job = history[prompt_id]

            if job.get("status", {}).get("status_str") == "error":
                raise HTTPException(
                    status_code=500,
                    detail=str(job)
                )

            break

    # Get generated image
    outputs = history[prompt_id].get("outputs", {})

    for node in outputs.values():

        for image in node.get("images", []):

            url = (
                f"{COMFY}/view"
                f"?filename={urllib.parse.quote(image['filename'])}"
                f"&subfolder={urllib.parse.quote(image['subfolder'])}"
                f"&type={urllib.parse.quote(image['type'])}"
            )

            try:
                image_bytes = urllib.request.urlopen(url).read()

                return image_bytes

            except Exception as e:
                raise HTTPException(
                    status_code=500,
                    detail=f"Failed retrieving generated image: {str(e)}"
                )

    raise HTTPException(
        status_code=500,
        detail="ComfyUI finished but returned no image"
    )


@app.get("/")
def root():
    return {
        "status": "online",
        "service": "Viggle Qwen Image API"
    }


@app.post("/generate")
def generate(req: GenerateRequest):

    if not req.prompt.strip():
        raise HTTPException(
            status_code=400,
            detail="Prompt cannot be empty"
        )

    image_bytes = generate_image(req.prompt)

    return Response(
        content=image_bytes,
        media_type="image/png"
    )
'''

with open("/kaggle/working/app.py", "w") as f:
    f.write(app_code)

print("app.py updated successfully")
```
